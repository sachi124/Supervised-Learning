```python
import pandas as pd
import numpy as np

df = pd.read_csv("Titanic-Dataset.csv")
df.info()
df.head()
```

```python
# Check the duplicated value
df.duplicated()
```

```python
# Identify column datatypes

cat_col = [col for col in df.columns if df[col].dtype == 'object']
num_col = [col for col in df.columns if df[col].dtype != 'object']
print("Categorical Data:", cat_col)
print("Numerical Column:", num_col)
```

```python
# Count the unique ID Values in the Categorical Columns
df[cat_col].nunique()
```

```python
# Calculate the missing values as percentage
percentage = round((df.isnull().sum() / df.shape[0]) * 100,2)
print(f"The percentage of missing values is {percentage}")
```

```python
# Drop Data heavy or Irrelevant Columns
df1 = df.drop(columns=['Name', 'Ticket', 'Cabin'])
df1 = df.dropna(subset=['Embarked'], inplace=True)
df1['Age'] = df1['Age'].fillna(df1['Age'].mean())
```
