## One Hot Encoding In Machine Learning

### 1. Using Pandas
### Pandas provide the .get_dummies() function to perforn one-hot encoding on categorical columns

```python
import pandas as pd

data ={
    'Employee ID': [10,20,30,40,50],
    'Gender': ['M', 'M', 'F', 'M', 'F'],
    'Shift': ['Day', 'Day', 'Night', 'Day', 'Night'],
    'Remarks': ['Good', 'Nice', 'Great', 'Good', 'Great']
}

df = pd.DataFrame(data)

print("Original Data")
print(df)

encoded_df = pd.get_dummies(
    df,
    columns = ['Gender', 'Shift', 'Remarks'],
    drop_first = True
)

print("\nOne-Hot Encoded Data:")
encoded_df
```

### Using Scikit-learn library 
### OneHotEncoder function to convert categorical to binary columns for ML models
#### Automatically identify categories
#### useful for machine learning preprocessing
#### select_dtypes() helps select categorical columns automatically

```python
from sklearn.preprocessing import OneHotEncoder

data ={
    'Employee ID': [10,20,30,40,50],
    'Gender': ['M', 'M', 'F', 'M', 'F'],
    'Shift': ['Day', 'Day', 'Night', 'Day', 'Night'],
    'Remarks': ['Good', 'Nice', 'Great', 'Good', 'Great']
}

df1 = pd.DataFrame(data)
print("Original Data:")
print(df1)

categorical_columns = df.select_dtypes(include=['object']).columns

encoder = OneHotEncoder(sparse_output=False)

encoded_data = encoder.fit_transform(df[categorical_columns])

encoded_df = pd.DataFrame(
    encoded_data,
    columns = encoder.get_feature_names_out(categorical_columns)
)

final_df = pd.concat(
    [df.drop(columns=categorical_columns), encoded_df],
    axis = 1
)

print("\nOne-Hot Encoded Data:")
print(final_df)
```

```python
# Using another example of scikit-learn

data = {
    'Customer_ID': [101, 102, 103, 104, 105, 106],
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female', 'Male'],
    'Plan': ['Basic', 'Premium', 'Basic', 'Standard', 'Premium', 'Basic'],
    'Payment_Method': [
        'Cash',
        'Card',
        'Online',
        'Card',
        'Online',
        'Cash'
    ],
    'Churn': [0, 1, 0, 1, 1, 0]
}

df = pd.DataFrame(data)
# print(df)

categorical_columns = df.select_dtypes(include=['object']).columns

encoder = OneHotEncoder(sparse_output=False)

encoded_data = encoder.fit_transform(
    df[categorical_columns]
)

# create the dataframe
encoded_df = pd.DataFrame(
    encoded_data,
    columns=encoder.get_feature_names_out(categorical_columns)
)

final_df = pd.concat([df.drop(columns=categorical_columns), encoded_df], axis=1)
print(final_df)
```
