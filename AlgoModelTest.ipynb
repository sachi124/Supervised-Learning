```python
# Creating feature engineering functions like skill matching
def safe_ratio(matched_count, total_count):
    return matched_count / total_count if total_count else 0.0

def calculate_skill_match(student, job):
    student_skills = set(student['skills'])
    required_skills = set(job['required_skills'])
    preferred_skills = set(job['preferred_skills'])

    matched_required = student_skills.intersection(required_skills)
    matched_preferred = student_skills.intersection(preferred_skills)

    missing_required = required_skills - student_skills
    missing_preferred = preferred_skills - student_skills

    return {
        "required_skill_match": safe_ratio(len(matched_required), len(required_skills)),
        "preferred_skill_match": safe_ratio(len(matched_preferred), len(preferred_skills)),
        "matched_required": sorted(matched_required),
        "matched_preferred": sorted(matched_preferred),
        "missing_required": sorted(missing_required),
        "missing_preferred": sorted(missing_preferred)
    }

# Education matching
def calculate_education_match(student, job):
    return float(student['education'] in job['education_requirement'])

# location Matching
valley_locations = {"kathmandu", "lalitpur", "bhaktapur"}

def calculate_location_match(student, job):
    student_location = student['location'].lower()
    job_location = job['location'].lower()

    if student_location in job_location:
        return 1.0

    if student_location in valley_locations:
        return 1.0

    return 0.5


# Work_mode Matching
def calculate_work_mode_match(student, job):
    return float(job['work_mode'] in student['work_mode_preference'])

# Availability matching
def calculate_availability_match(student, job):
    if student['availability'] == 'full_time':
        return 1.0
    return 0.5

# project relevence
def calculate_project_relevance(student, job):
    project_text = " ".join(student["projects"]).lower()
    job_text = (
        job["title"] + " " +
        job["description"] + " " +
        " ".join(job["required_skills"]) + " " +
        " ".join(job["preferred_skills"])
    ).lower()

    relevant_terms = [
        "erp",
        "accounting",
        "finance",
        "testing",
        "data",
        "dashboard",
        "reporting",
        "documentation",
        "python",
        "sql",
        "excel",
        "business"
    ]

    matched_terms = [
        term for term in relevant_terms
        if term in project_text and term in job_text
    ]

    return min(len(matched_terms) / 4, 1.0)
```

```python
# TF_IDF for text similarity
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def calculate_text_similarity(student, job):
    student_text = (
        student["resume_text"] + " " +
        " ".join(student["skills"]) + " " +
        " ".join(student["projects"])
    )

    job_text = (
        job["title"] + " " +
        job["description"] + " " +
        " ".join(job["required_skills"]) + " " +
        " ".join(job["preferred_skills"])
    )

    vectorizer = TfidfVectorizer(
        lowercase=True,
        stop_words="english",
        ngram_range=(1, 2)
    )

    vectors = vectorizer.fit_transform([
        student_text,
        job_text
    ])

    return float(
        cosine_similarity(
            vectors[0:1],
            vectors[1:2]
        )[0][0]
    )

# TF-IDF transforms the text into vectors, while cosine similarity measures the similarity between the student and role vectors
```

```python
# build one KNN features vector
def build_features(student, job):
    skill_result = calculate_skill_match(student, job)

    required_score = skill_result["required_skill_match"]

    experience_score = min(
        0.30 +
        (0.70 * required_score),
        1.0
    )

    return {
        "required_skill_match": required_score,
        "preferred_skill_match": skill_result[
            "preferred_skill_match"
        ],
        "text_similarity": calculate_text_similarity(student, job),
        "education_match": calculate_education_match(student, job),
        "location_match": calculate_location_match(student, job),
        "work_mode_match": calculate_work_mode_match(student, job),
        "availability_match": calculate_availability_match(student, job),
        "project_relevance": calculate_project_relevance(student, job),
        "matched_required": skill_result["matched_required"],
        "matched_preferred": skill_result["matched_preferred"],
        "missing_required": skill_result["missing_required"],
        "missing_preferred": skill_result["missing_preferred"]
    }
```

```python
# Train the KNN
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

training_df = pd.DataFrame(training_examples)

feature_columns = [
    "required_skill_match",
    "preferred_skill_match",
    "text_similarity",
    "education_match",
    "location_match",
    "work_mode_match",
    "availability_match",
    "project_relevance"
]

X_train = training_df[feature_columns]
y_train = training_df["label"]

knn_model = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier(
        n_neighbors=3,
        weights="distance"
    ))
])

knn_model.fit(X_train, y_train)
```

```python
# Predict the student
def predict_student_for_job(student, job, model):
    features = build_features(student, job)

    input_df = pd.DataFrame([{
        column: features[column]
        for column in feature_columns
    }])

    prediction = int(model.predict(input_df)[0])

    probability = float(
        model.predict_proba(input_df)[0][1]
    )

    if probability >= 0.75:
        recommendation = "Strong recommendation"
    elif probability >= 0.55:
        recommendation = "Review recommended"
    else:
        recommendation = "Weak match"

    return {
        "student_id": student["student_id"],
        "student_name": student["name"],
        # "job_id": job["job_id"],
        "job_title": job["title"],
        "prediction": prediction,
        "suitability_score": round(probability * 100, 2),
        "recommendation": recommendation,
        "matched_required": features["matched_required"],
        "missing_required": features["missing_required"],
        "matched_preferred": features["matched_preferred"],
        "missing_preferred": features["missing_preferred"]
    }
```

```python
# Final answer
result = predict_student_for_job(
    students[0],
    jobs[0],
    knn_model
)

result
```

```python
def recommend_students_for_job(job, students, model):
    recommendations = []

    for student in students:
        result = predict_student_for_job(
            student,
            job,
            model
        )

        # Hard filter: do not strongly recommend someone
        # who matches fewer than 30% of required skills.
        required_match = (
            len(result["matched_required"]) /
            len(job["required_skills"])
        )

        if required_match < 0.30:
            result["recommendation"] = (
                "Low eligibility - manual review only"
            )

        recommendations.append(result)

    return sorted(
        recommendations,
        key=lambda item: item["suitability_score"],
        reverse=True
    )
```

```python
functional_recommendations = recommend_students_for_job(
    jobs[0],
    students,
    knn_model
)

for candidate in functional_recommendations:
    print(
        candidate["student_name"],
        "-",
        candidate["suitability_score"],
        "%",
        "-",
        candidate["recommendation"]
    )
```
